# Bestimmtheitsmaß R² Score (Determinationskoeffizient)

* Berechne den Fehler für ein "richtig schlechtes" Modell
* Wie viel besser ist unser Modell als das "richtig schlechte" Modell

"Richtig schlecht" => z.B. ein Modell welches nur den Durchschnitt des y Werts berücksichtigt (bei Karat/Preis Beispiel Karat einfach ignorieren)

* Wert 1 = Modell, dass die Daten perfekt beschreibt
* Wert 0 = Modell, dass die Daten genauso gut wie das "richtig schlechte" beschreibt
* Wert < 0 = Modell, dass die Daten noch schlechter beschreibt

Die Formel für das Bestimmtheitsmaß (R²) lautet:

$$
R^2 =  \frac{\sum_{i=1}^n (\hat{y}_i - \bar{y}_i)^2}{\sum_{i=1}^n (y_i - \bar{y})^2}
$$

- $y_i$: Tatsächlicher Wert
- $\hat{y}_i$: Vorhergesagter Wert
- $\bar{y}$: Mittelwert der tatsächlichen Werte
