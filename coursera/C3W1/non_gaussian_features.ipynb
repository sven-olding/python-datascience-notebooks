```python
from scipy.stats import skewnorm
import matplotlib.pyplot as plt
import numpy as np

num_values = 10_000
max_value = 100
skewness = 20

random = skewnorm.rvs(a=skewness, loc=max_value, size=num_values)

random = random - min(random) # Shift the distribution to start at 0
random = random / max(random) # Scale the distribution to fit within the range of 0 to 1
random = random * max_value # Scale the distribution to fit within the range of 0 to max_value

x = random
plt.hist(x, bins=50, density=True)
```

```python
plt.hist(x**0.5, bins=50, density=True, alpha=0.5, label='Square Root')
```

```python
plt.hist(np.log(x + 8), bins=50, density=True, alpha=0.5, label='Logarithm')
```
