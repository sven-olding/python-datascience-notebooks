```python
import numpy as np

AT = np.array([[200, 17]]) # 1x2

W = np.array([[1, -3, 5],
              [-2, 4, -6]]) # 2x3

B = np.array([-1, 1, 2]) # 1x3

Z = AT.dot(W) + B  # 1x3

print(Z)
```

```python

def dense(AT, W, B, activation): # in TensorFlow AT would be a_in by convention
    z = np.matmul(AT, W) + B
    return activation(z) # Activation function to be passed as an argument
    
```
