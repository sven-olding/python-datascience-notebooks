```python
import tensorflow as tf
```

```python
# Auto Diff
w = tf.Variable(3.0)    # tf.Variables are the parameters we want to optimize. They are initialized to some value, in this case 3.0

x = 1.0
y = 1.0 # target value

alpha = 0.01

iterations = 30

for iter in range(iterations):
    # Use gradient tape to record the steps used to compute the cost J to enable auto differentiation
    
    # The cost function is the squared error between the predicted value f(x) and the target value y, which is (f(x) - y) ** 2. In this case, f(x) is simply w * x.
    with tf.GradientTape() as tape:
        fwb = w * x # f(x)
        cost_j = (fwb - y) ** 2
    
    # Use the gradient tape to compute the gradient of cost_j with respect to w    
    # this will find the derivative of cost_j with respect to w, which is dJ/dw, and store it in the variable dJdw
    [dJdw] = tape.gradient(cost_j, [w])
    
    # Run one step of gradient descent on w using the learning rate alpha and the gradient dJdw
    w.assign_add(-alpha * dJdw)
```
